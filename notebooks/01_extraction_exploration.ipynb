# 01 — Extraction Exploration

Explore PDF text extraction from 10-K filings.

```python
import random

# For each extracted file, look at random samples
def inspect_extraction(filename, n_samples=3):
    with open(f"E:/P1_JOB/AI projects/financial_risk_rag/data/extracted_text/{filename}", 'r', encoding='utf-8') as f:
        text = f.read()
    
    print(f"=" * 60)
    print(f"INSPECTING: {filename}")
    print(f"Total characters: {len(text):,}")
    print(f"=" * 60)
    
    # Look at start, middle, end, plus 2 random spots
    positions = [
        ("Start", 0),
        ("25% in", len(text) // 4),
        ("Middle", len(text) // 2),
        ("75% in", 3 * len(text) // 4),
        ("End", len(text) - 500)
    ]
    
    for label, pos in positions:
        print(f"\n--- {label} (chars {pos}–{pos+500}) ---")
        print(text[pos:pos+500])
```

```python
inspect_extraction('AXP_0000004962-26-000080_risk_factors.txt')
```
