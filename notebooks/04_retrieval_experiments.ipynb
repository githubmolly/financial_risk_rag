# 03 — Retrieval Experiments

Test retrieval quality: embedding models, similarity metrics, top-k.

```python
import chromadb
from sentence_transformers import SentenceTransformer


client = chromadb.PersistentClient(path="E:\\P1_JOB\\AI projects\\financial_risk_rag\\chroma_db")
collection = client.get_collection("financial_risk_10k")
model = SentenceTransformer("BAAI/bge-small-en-v1.5")
print(f"Total chunks: {collection.count()}")
# Test query
query = "What are the main credit risks facing the company?"
query_emb = model.encode(query, normalize_embeddings=True).tolist()
results = collection.query(query_embeddings=[query_emb], n_results=5)
for i in range(5):
    print(f"\n#{i+1} from {results['metadatas'][0][i]['source']}")
    print(f"Section: {results['metadatas'][0][i].get('item', 'unknown')}")
    print(f"   {results['documents'][0][i][:200]}...")
```
