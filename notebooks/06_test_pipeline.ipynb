```python
# Cell 1 — Imports with defensive env loading
import os
from pathlib import Path
from dotenv import load_dotenv

# Try multiple plausible locations for .env
candidate_paths = [
    PROJECT_ROOT / ".env",
    Path.cwd() / ".env",
    Path("E:/P1_JOB/AI projects/financial_risk_rag/.env"),
]

loaded_from = None
for path in candidate_paths:
    if path.exists():
        load_dotenv(path)
        loaded_from = path
        break

if loaded_from is None:
    raise FileNotFoundError(
        f"No .env file found in any of: {[str(p) for p in candidate_paths]}"
    )

# Verify the key actually loaded
if not os.getenv("GEMINI_API_KEY"):
    raise ValueError(
        f"Loaded .env from {loaded_from} but GEMINI_API_KEY is empty. "
        f"Check the file contents — it should have a line like "
        f"GEMINI_API_KEY=AIza... with no quotes or spaces."
    )

print(f"✓ Loaded .env from {loaded_from}")
print(f"✓ GEMINI_API_KEY length: {len(os.getenv('GEMINI_API_KEY'))} chars")

# Now the rest of your imports
from src.retrieval import HybridRetriever
from src.generation import Generator
from src.pipeline import RAGPipeline
```
